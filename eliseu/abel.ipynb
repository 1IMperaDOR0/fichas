# 📖 HISTÓRIA

Ishgard. Antes conhecido como “o continente esquecido”, foi salvo e restaurado pelas mãos da Guardiã Erika e seus companheiros Aventurine, Dian, Greed, Ko-rinn, M7 e Vicky, na batalha contra o Salvador.

Diferente dos grandes impérios e reinos antigos do mundo, Ishgard é recente, um continente ainda em desenvolvimento. Um lugar onde as cidades ainda não nasceram, e as histórias estão apenas começando a ser escritas. Aqui, entre a natureza bruta e a energia divina instável, surgiram pequenas vilas. E entre elas, uma em especial: a vila dos anjos menores puros.

Um refúgio sagrado, lar de seres celestiais com dom inato para a cura. Ali, entre rebanhos de ovelhas, plantações de ervas sagradas e relíquias mágicas deixadas por viajantes em troca de cuidados, nasceu Eliseu Babosa.

Desde pequeno, Eliseu aprendeu o valor de proteger.

Mas também aprendeu o que é a solidão.

Ele foi um dos últimos anjos puros daquela geração. Sem outras crianças da sua idade, cresceu entre adultos ocupados e anciãos silenciosos. Seus únicos amigos eram as ovelhas.

Era solitário, mas sempre prestativo. Havia sempre algo a fazer: um ferimento para cuidar, uma erva para colher, um rebanho para alimentar. Fazia tudo com dedicação.

Mas, em seu coração, havia uma inquietação constante:

> “Eu queria poder dividir isso com alguém.”

Eliseu sonhava com um rebanho só seu.

Não de ovelhas.

De amigos.

Pessoas em quem pudesse confiar. Pessoas que também confiassem nele. E, por elas, seria capaz de tudo.

Proteger era mais do que um instinto.

Era seu sonho.

Na vila, era tradição: ao completar dezesseis anos, cada anjo partia para a Escola de Magia, onde aprenderia a canalizar seu dom e se tornaria um guardião da integridade no mundo exterior.

E assim, Eliseu partiu.

Não apenas para aprender magia.

Mas para buscar aquilo que lhe faltava.

Amigos.

Parceiros.

Meio igual a meio.

E alguém por quem valesse a pena lutar.

## 🏫 ARCÁDIA

A chegada à Escola de Magia de Arcádia foi o primeiro grande passo de Eliseu para fora da pequena vila onde havia crescido.

Pela primeira vez, ele estava cercado por pessoas da sua idade.

E, pela primeira vez, teve a oportunidade de construir aquilo que sempre desejou.

Foi ali que conheceu Aurélia, Orianna, Matteo e Ariarali.

Quatro pessoas que, inicialmente, eram apenas colegas.

Com o tempo, porém, tornaram-se muito mais do que isso.

Tornaram-se seu grupo.

Seus companheiros.

Seu novo rebanho.

A escola, entretanto, não estava interessada em facilitar suas vidas.

Logo no início, os cinco foram submetidos a um exame de sobrevivência. O objetivo era entrar em uma caverna e descobrir, por conta própria, o verdadeiro significado da magia.

Tudo isso para satisfazer os desejos dos cinco egocêntricos que comandavam aquela prova.

Não era simplesmente uma avaliação de conhecimento.

Era uma prova para descobrir se aqueles jovens seriam capazes de sobreviver ao mundo que existia além dos muros de Arcádia.

E eles sobreviveram.

Mas, dentro daquela caverna, encontraram algo que mudaria seus destinos para sempre.

Uma mulher misteriosa.

Uma cabana.

E um baralho de Tarot.

Cada um deles recebeu uma carta.

Não eram cartas comuns. Eram artefatos capazes de despertar poderes ocultos em seus portadores, poderes que nenhum deles compreendia completamente naquele momento.

As cartas escolhidas foram:

**A Torre.**

**O Diabo.**

**A Imperatriz.**

**O Enforcado.**

**O Julgamento.**

A partir daquele momento, suas vidas passaram a estar ligadas a algo muito maior do que a própria escola.

Eles ainda não sabiam.

Mas aquelas cartas eram convites.

Convites para uma caça.

Uma caça pelos pilares da própria realidade.

## ✨ OS ANOS EM ARCÁDIA

Os anos seguintes foram uma mistura de aprendizado, amizade, perigo e acontecimentos que nenhum professor deveria considerar normais dentro de uma escola de magia.

Eles estudaram com Naxa, considerada por muitos a melhor professora de Arcádia.

Aprenderam magia.

Conheceram novos amigos.

Treinaram seus corpos.

Fizeram academia.

Enfrentaram maldições.

Participaram de campeonatos de culinária.

E, em uma ocasião particularmente memorável, Eliseu decidiu que a melhor maneira de compreender sua carta de Tarot seria jogar uma antibíblia em uma escada.

A escada não sobreviveu.

Eliseu também não ganhou exatamente a resposta que procurava.

Mas, de alguma maneira, continuou estudando.

A escola também os colocou diante de ameaças muito mais sérias.

Em determinado momento, eles descobriram um esquema envolvendo o sequestro de crianças. O grupo enfrentou os responsáveis e resistiu até onde conseguiu.

Não foi suficiente.

Eles morreram.

Todos.

Mas sua história ainda não terminaria ali.

Topázio, conhecida por seu coração de pedra, interveio.

Com a ajuda de sua assistente, que possuía o poder de ressuscitar os mortos, os cinco retornaram à vida.

A comemoração que veio depois provavelmente não deveria ser chamada de comemoração.

Havia bebidas.

Comida.

Coxinhas.

E coisas que definitivamente não deveriam estar naquela festa.

Mas, para eles, estar vivos novamente já era motivo suficiente para comemorar.

## 🗼 A TORRE

No meio do ano, outro grande evento surgiu.

Uma torre.

Diversos grupos.

Um objetivo simples:

**alcançar o topo.**

Para isso, os grupos precisavam enfrentar uns aos outros.

A princípio, todos acreditavam que aquilo significava matar seus adversários.

Mas descobriram que a “morte” não era exatamente literal.

Ainda assim, a competição foi brutal.

E, enquanto todos lutavam para subir, Eliseu parecia ter encontrado a estratégia mais eficiente de todas:

**sobreviver por pura sorte.**

Entre todos os acontecimentos daquela competição, porém, um feito merece ser registrado.

Matteo matou brutalmente quatro pessoas a sangue frio.

Quatro.

Isso, naturalmente, tornou-se uma informação impossível de esquecer.

Ou de deixar de usar contra ele.

## 🏹 A CAÇA

Por trás de todas aquelas aventuras existia algo muito maior.

As cartas de Tarot não eram apenas fontes de poder.

Eram uma oportunidade.

Aqueles que as possuíam poderiam participar de uma caça pelos pilares da realidade.

Pilares que sustentavam o próprio mundo.

Se fossem destruídos ou corrompidos, as consequências poderiam atingir tudo o que existia.

O grupo aceitou o fardo.

E, entre todos os pilares, um deles se tornou especialmente importante:

**o Pilar do Tempo.**

Eles encontraram seu portador durante uma de suas jornadas.

Ele era um viajante do tempo.

Mas estava doente.

Instável.

A princípio, parecia que estavam diante do próprio Pilar.

Mas logo descobriram a verdade.

Ele não era o Pilar.

Era apenas seu portador.

Assim como alguém pode carregar dentro de si uma força monstruosa sem ser a própria criatura, aquele viajante carregava uma parte daquele poder.

O grupo decidiu ajudá-lo.

Com os poderes concedidos pelas cartas, conseguiram estabilizá-lo.

E, durante esse processo, fizeram amizade com ele.

A caça, porém, começou a cobrar seu preço.

## ⏳ A ORIANNA DO FUTURO

Algum tempo depois, algo impossível aconteceu.

Orianna voltou do futuro.

Sua carta estava corrompida.

Ela conhecia o destino que aguardava aquele grupo.

E tentou impedir que seus amigos continuassem.

Seu aviso era simples:

**abandonem a caça.**

Mas eles não aceitaram.

Orianna insistiu.

Tentou convencê-los.

Mas aqueles que haviam aceitado o fardo não estavam dispostos a abandoná-lo.

Então aconteceu o inevitável.

Orianna do futuro enfrentou seus próprios amigos.

Eliseu e Ariarali estavam dormindo quando o confronto começou.

Os outros três precisaram enfrentá-la.

E conseguiram derrotá-la.

Mas, antes de desaparecer, Orianna deixou uma última mensagem.

Uma advertência.

Ela havia visto as consequências.

Sabia o que aconteceria.

E deixou apenas uma certeza:

> “A morte está vindo.”

Eles não deram ouvidos.

Ou talvez tenham dado.

Talvez apenas tenham decidido continuar mesmo assim.

## ⚔️ O FIM DA ESCOLA

Os anos passaram.

As crianças que chegaram a Arcádia tornaram-se adultos.

Aprenderam magia.

Lutaram.

Perderam pessoas.

Fizeram amigos.

Cometeram erros.

E sobreviveram a coisas que nenhum estudante deveria precisar enfrentar.

Até que chegou o momento de partir.

A escola havia terminado.

Mas a verdadeira missão deles estava apenas começando.

Era hora de concluir a caça.

Para alcançar os pilares, precisaram atravessar uma região que mais parecia uma guerra.

Explosões.

Mortes.

Criaturas.

Magia.

Caos.

Cada passo adiante exigia uma luta pela própria sobrevivência.

No caminho, encontraram o campeão do Leviatã.

Uma criatura ligada a uma força que também precisava ser destruída.

O combate foi brutal.

Mas o grupo venceu.

O campeão caiu.

E, finalmente, eles conseguiram alcançar a região onde os pilares poderiam ser encontrados.

Exaustos, feridos e quase sem forças, chegaram a uma cidade que decidiu acolhê-los.

Ali, pela primeira vez em muito tempo, puderam descansar.

Dormiram.

E, no dia seguinte, fizeram aquilo que sabiam fazer melhor:

planejar como sobreviver à próxima coisa impossível.

Cada pilar seria enfrentado de uma maneira diferente.

Cada integrante teria uma função.

Cada possibilidade havia sido considerada.

Eles estavam preparados.

Ou acreditavam estar.

## 🩸 A MORTE DE ELISEU

Talvez Orianna estivesse certa.

Talvez todos os avisos tivessem sido verdadeiros.

Talvez aquele fosse o destino que jamais poderia ser evitado.

Depois de tantos anos juntos, de tantas aventuras, de tantas batalhas e de tudo aquilo que haviam enfrentado lado a lado...

o destino finalmente alcançou Eliseu.

E, ironicamente, não foi um monstro.

Não foi um pilar.

Não foi uma maldição.

Foi alguém em quem ele confiava.

**Aurélia.**

Foi pelas mãos dela que o assassinato de Eliseu finalmente se concretizou.

O rapaz que havia saído de uma pequena vila procurando amigos.

O anjo que sonhava em proteger aqueles que amava.

O jovem que queria apenas encontrar seu próprio rebanho.

Morreu pelas mãos de alguém que fazia parte dele.

## 🕯️ O ÚLTIMO SUSPIRO

A morte deveria ser o fim.

Mas, para Eliseu, não foi.

Nos últimos suspiros de sua alma, quando sua existência já deveria estar desaparecendo, alguém o aguardava.

Nefariel.

Uma entidade que lhe ofereceu aquilo que ninguém mais poderia oferecer:

**uma segunda chance.**

Mas Nefariel não estava oferecendo um presente.

Estava oferecendo um contrato.

Uma vida em troca de um preço.

Um propósito em troca de lealdade.

Uma nova existência em troca de servir à sua causa.

Eliseu havia perdido tudo.

Seus amigos.

Sua casa.

Seu futuro.

Sua chance de voltar para a vila.

E, principalmente, a confiança nas pessoas pelas quais havia lutado durante tantos anos.

Disseram a ele que não havia mais volta.

Que não havia sobrevivência.

Que tudo aquilo pelo que haviam lutado talvez tivesse sido em vão.

E, naquele momento, Eliseu precisou tomar uma decisão.

Voltar para o nada...

ou aceitar a mão que lhe era estendida.

Ele escolheu viver.

Mas não como Eliseu.

Não mais como o anjo que partiu de sua vila aos dezesseis anos.

Nefariel lhe deu um novo nome.

**Abel.**

E, com esse novo nome, nasceu também uma nova pessoa.

Eliseu Babosa morreu naquela batalha.

O que voltou não era exatamente o mesmo homem.

Era alguém que carregava suas memórias, suas dores e seus arrependimentos.

Alguém que havia aprendido da pior maneira possível que proteger os outros não significava necessariamente ser protegido por eles.

E agora, pela primeira vez em sua vida, ele não queria encontrar um rebanho.

Não queria voltar para casa.

Não queria ser um guardião.

Ele queria fazer a causa de Nefariel se tornar realidade.

Porque, depois de tudo que havia acontecido, Abel já não sabia mais se estava lutando para salvar o mundo...

ou para destruí-lo.

### **INVOCAÇÕES MÍSTICAS (Nível 1)**

Você tem as seguintes invocações místicas:

* **Lança Mística:** Escolha um dos seus truques conhecidos de Bruxo que cause dano e tenha um alcance de 3 metros ou mais. Ao conjurar essa magia, seu alcance aumenta em um número de metros igual a 9 vezes o seu nível de Bruxo

* **Mente Mística:** Você tem Vantagem em salvaguardas de Constituição que realiza para manter a Concentração.

* **Pacto da Corrente:** Você aprende a magia Convocar Familiar e pode conjurá-la como uma ação Usar Magia sem gastar um espaço de magia.

    Ao conjurar a magia, você escolhe uma das formas normais para o seu familiar ou uma das seguintes formas especiais: Cobra Peçonhenta, Diabrete, Esfinge Maravilhosa, Esqueleto, Pseudodragão, Quasit, Slaad Girino, Sprite (veja o apêndice B para o bloco de estatísticas do familiar).

    Além disso, ao executar ação Atacar, você pode renunciar a um de seus próprios ataques para permitir que seu familiar realize um ataque próprio como uma Reação dele.

* **Investimento do Mestre da Corrente:** Ao conjurar Convocar Familiar, você infunde o familiar invocado com uma dose de seu poder místico, concedendo à criatura os seguintes benefícios.

    Aéreo ou Aquático. O familiar obtém um Deslocamento de Voo ou um Deslocamento de Natação (à sua escolha) de 12 metros.

    Ataque Rápido. Como uma Ação Bônus, você pode ordenar o familiar a executar a ação Atacar.

    CD para Evitar sua Magia. Se o familiar forçar uma criatura a realizar uma salvaguarda, ela usa a CD para evitar sua magia.

    Dano Necrótico ou Radiante. Sempre que o familiar causar dano Contundente, Cortante ou Perfurante, você pode substituir o dano por Necrótico ou Radiante.

    Resistência. Quando o familiar sofre dano, você pode executar uma Reação para conceder Resistência contra esse dano.

* **Explosão Repulsiva:** Escolha um dos seus truques conhecidos de Bruxo que exija uma jogada de ataque. Ao atingir uma criatura Grande ou menor com esse truque, você pode empurrá-la até 3 metros para longe de você.

* **Pacto do Manto:** Como uma Ação Bônus, você pode invocar um manto, capa ou robe que auxilia seu movimento e mobilidade.

    Enquanto estiver usando o manto:

    Você flutua alguns centímetros acima do chão.
    Você ignora os efeitos de terreno difícil.
    Seu deslocamento aumenta em 3 metros.
    Você pode caminhar sobre superfícies líquidas, como água e areia movediça.
    O manto não impede seu movimento, visão ou utilização de outros equipamentos.

    Você pode dispensar o manto como uma Ação Bônus.

### **FLUXO PRECISO**

Fluxo representa precisão, Foco e esforço e a base para a conjuração 

Nível 0 - Você aprende a magia Assassino fantasmagórico pode conjurar ela 1 vez por dia sem usar espaços de magia 

Nível 4 - Foco primário: Suas magias com alvo único tem a CD aumentada em +6 e causam 2 dados extras de dano 

Nível 8 - 1 por um: Você pode reduzir a área de magias a sua vontade, se fizer isso, o dano aumenta em 4 dados para cada 3m reduzidos da área inicial 

Nível 12 - Derramamento de sangue: Causar dano com magias de alvo único causa +2 de dano por dado 

Nível 16 - Presa fácil: Causar dano a um inimigo sem aliados em um alcance de 3m aumenta a categoria do dado em 1, se seus aliados também não estiverem nesse alcance a categoria aumenta em 3 invés de 1

Nível 20 - Caça perfeita: A primeira magia de alvo único que você conjurar no turno não gasta nenhuma ação para ser conjurada

Essa característica transforma contratos mágicos, pactos e acordos realizados pelo personagem em efeitos potencialmente permanentes, dependendo das condições estabelecidas no próprio contrato.

### **MÃOS CURATIVAS (Raça)**

Você executa uma ação Usar Magia, toca uma criatura e joga um número de d4s igual ao seu Bônus de Proficiência. A criatura restaura número de Pontos de Vida igual ao total jogado. Após usar esse traço, você não pode usá-lo novamente até completar um Descanso Longo.

---

### **PORTADOR DA LUZ (Raça)**

Você conhece o truque Luz. Carisma é seu atributo de conjuração para isso.

---

### **REVELAÇÃO CELESTIAL (Raça)**

No nível 3 de personagem, você pode se transformar como uma Ação Bônus usando uma das opções abaixo (escolha a opção cada vez que você se transformar). A transformação se mantém por 1 minuto ou até você a encerrar (nenhuma ação é necessária). Uma vez que você se transforma, não pode fazê-lo novamente até completar um Descanso Longo. Uma vez em cada um dos seus turnos, até que a transformação termine, você pode infligir dano adicional a um alvo ao causar dano a ele com um ataque ou uma magia. O dano adicional é igual ao seu Bônus de Proficiência, e o tipo de dano adicional é Necrótico para Manto Necrótico ou Radiante para Asas Celestiais e Transfiguração Radiante. Aqui estão as opções de transformação:

* **Asas Celestiais:** Duas asas espectrais brotam em suas costas temporariamente. Até que a transformação se encerre, você tem um Deslocamento de Voo igual ao seu Deslocamento.

* **Manto Necrótico:** Seus olhos se tornam brevemente poças de escuridão, e asas que não voam brotam em suas costas temporariamente. Criaturas que não sejam seus aliados a até 3 metros de você devem ser bem-sucedidas em uma salvaguarda de Carisma (CD 8 + seu modificador de Carisma e seu Bônus de Proficiência) ou têm a condição Amedrontado até o final do seu próximo turno.

* **Transfiguração Radiante:** Luz abrasadora irradia temporariamente de seus olhos e boca. Pela duração da transformação, você emite Luz Plena em um raio de 3 metros e Meia-luz por mais 3 metros, e no fim de cada um de seus turnos, cada criatura a até 3 metros de você sofre dano Radiante igual ao seu Bônus de Proficiência

### **VIGOROSO (Origem)**

Você ganha 2 de vida para cada nível de personagem.

---

### **SORTUDO (Origem)**

* **Pontos de Sorte:** Você tem um número de Pontos de Sorte igual ao seu Bônus de Proficiência e pode gastar os pontos nos benefícios abaixo. Você restaura seus Pontos de Sorte gastos quando completa um Descanso Longo.
* **Vantagem:** Quando você joga um d20 para um Teste de D20, pode gastar 1 Ponto de Sorte para ter Vantagem na jogada.
* **Desvantagem:** Quando uma criatura jogar um d20 para uma jogada de ataque contra você, você pode gastar 1 Ponto de Sorte para impor Desvantagem nessa jogada
