## 📖 **HISTÓRIA**

Muito antes da existência de reinos, impérios e cidades, quando o mundo ainda pertencia às grandes florestas, aos rios selvagens e às criaturas colossais, pequenas tribos disputavam diariamente o direito de sobreviver. Entre elas existia uma comunidade diferente de todas as outras.

Enquanto a maioria dos povos vivia isolada e desconfiava de qualquer raça que não fosse a sua, dois grupos exóticos perceberam que, separados, caminhavam lentamente para a extinção. Assimars e draconatos decidiram fazer algo que, para aquela época, parecia impossível: unir suas tribos.

Não foi uma aliança política.

Não existiam reis.

Não existiam chefes.

Nem sequer havia um conceito de nação.

Era simplesmente uma comunidade formada por indivíduos que entenderam que sobreviver juntos era melhor do que desaparecer separados.

Com o passar das gerações, assimars, draconatos e seus descendentes híbridos começaram a nascer em grande número. Embora utilizassem uma linguagem extremamente primitiva, já eram capazes de transmitir ideias, valores, crenças e tradições. Dividiam alimento, protegiam uns aos outros e aprendiam juntos a enfrentar um mundo extremamente hostil.

Assim nasceu a Tribo dos Exóticos.

Foi nela que nasceu Mim Mace Tartu.

Desde pequeno, Mim demonstrava uma força incomum até mesmo entre seu próprio povo. Cresceu muito mais alto e musculoso do que qualquer outro membro da tribo, resultado de uma rara mutação que ninguém sabia explicar. Alguns acreditavam que era uma bênção dos deuses. Outros diziam que era apenas uma peculiaridade da natureza.

Para Mim, isso nunca importou.

A força não existia para dominar.

Existia para proteger.

Servo fiel de Tharuun, o conceito da natureza, ele acreditava que toda forma de vida fazia parte de um grande ciclo. Nunca caçava por prazer ou diversão. Toda presa abatida era seguida por uma breve oração de agradecimento, desejando que seu espírito encontrasse paz e retornasse à natureza para dar origem a uma novo ciclo.

Nada era desperdiçado.

A carne alimentava a tribo.

Os ossos tornavam-se ferramentas.

O couro virava roupas.

Até mesmo o sangue era devolvido à terra como forma de respeito.

Para Mim, matar sem necessidade era um insulto ao próprio ciclo da vida.

Os anos passaram e a tribo prosperou.

Duas grandes descobertas transformaram completamente aquele povo.

A primeira foi aprender que algumas sementes poderiam ser plantadas e cultivadas, permitindo que a comida crescesse perto de casa.

A segunda foi ainda mais surpreendente.

Alguns dos grandes répteis herbívoros que antes serviam apenas como alimento podiam conviver ao lado dos caçadores.

Entre os primeiros filhotes domesticados existia um que ninguém conseguia controlar.

Era maior do que os demais.

Mais forte.

Mais agressivo.

Rosnava, mordia e investia contra qualquer um que tentasse se aproximar.

Todos desistiram dele.

Menos Mim.

Durante muitos dias, ele apenas se aproximava lentamente, deixando comida e água sem tentar prendê-lo ou obrigá-lo a obedecer. Sentava-se próximo ao animal e permanecia em silêncio, permitindo que a confiança surgisse naturalmente.

Com o tempo, a desconfiança desapareceu.

Depois veio a curiosidade.

E, por fim, nasceu uma amizade.

Mim deu ao grande réptil o nome de Mui.

Na antiga língua da tribo, esse nome significava algo próximo de "aquele que caminha ao meu lado".

Os dois tornaram-se inseparáveis.

Mui transportava Mim durante as caçadas, cruzava rios ao seu lado e enfrentava predadores sem hesitação. Com o passar dos anos, deixaram de agir como montaria e cavaleiro.

Lutavam como se fossem uma única criatura.

Enquanto a tribo começava a sonhar com um futuro de prosperidade, Mim alcançava a idade adulta. Em breve escolheria uma companheira para formar uma família e ajudar a fortalecer sua comunidade.

Mas esse futuro nunca aconteceu.

Os humanos cresciam em um ritmo assustador.

Sua população aumentava rapidamente, e sua ambição era ainda maior.

Ao descobrirem a prosperidade da Tribo dos Exóticos, passaram a enxergá-la como uma ameaça ao seu domínio sobre aquelas terras.

Guiados pelos seguidores de Brask, marcharam para a primeira grande guerra entre tribos.

Foi um massacre.

A floresta queimou.

Os rios ficaram vermelhos.

Criaturas fugiram em todas as direções.

Guerreiros que até então lutavam apenas para sobreviver passaram a lutar para impedir a própria extinção.

Mim combateu como nunca antes.

Com seu enorme glave em mãos e Mui ao seu lado, derrubou inúmeros inimigos enquanto protegia seus companheiros.

Mas nem mesmo sua força foi suficiente para impedir a tragédia.

Após horas de batalha, nenhum dos lados possuía forças para continuar.

Os humanos haviam perdido incontáveis guerreiros.

Os Exóticos haviam perdido parte de sua própria história.

Sem condições de prosseguir, ambos recuaram.

Coberto de sangue, exausto e acreditando que finalmente voltaria para sua tribo...

...o próprio mundo pareceu se partir.

Uma energia desconhecida rasgou o espaço ao seu redor.

O céu desapareceu.

O chão deixou de existir.

O tempo pareceu parar.

Instintivamente, Mui correu em sua direção.

Sem pensar.

Sem hesitar.

Saltou para alcançá-lo no exato instante em que a estranha fenda consumia Mim.

Os dois desapareceram juntos.

Quando Mim voltou a abrir os olhos, a guerra havia desaparecido.

Sua tribo não estava mais ali.

Não havia rastros dos humanos.

Nem das florestas que conhecia.

Nem de sua casa.

Tudo o que restava era um mundo completamente desconhecido...

...e Mui, ainda ao seu lado.

Mim não compreendia quem o havia levado até aquele lugar.

Não sabia se sua tribo sobrevivera.

Nem se algum dia voltaria para casa.

Mas uma certeza permanecia inabalável.

Enquanto respirasse, continuaria seguindo os ensinamentos de Tharuun.

Protegeria aqueles que caminhassem ao seu lado.

Respeitaria o ciclo da vida.

E enfrentaria qualquer ameaça que tentasse romper o equilíbrio da natureza.

Pois, para Mim Mace Tartu, a verdadeira força nunca esteve em destruir.

Sempre esteve em preservar aquilo que vale a pena proteger.

### **FÚRIA (Nível 1)**

Como uma Ação Bônus, você entra em um estado de fúria que dura 1 minuto.

Enquanto estiver em fúria:

* Vantagem em testes de Força e testes de resistência de Força
* Bônus de dano em ataques Corpo a Corpo com Força
* Resistência a dano contundente, perfurante e cortante

**Bônus de dano da Fúria:**

* Nível 1–8: +2
* Nível 9–15: +3
* Nível 16–20: +4

A fúria termina se você:

* não atacar ou não sofrer dano no turno
* cair inconsciente

---

### **DEFESA SEM ARMADURA (Nível 1)**

Enquanto não estiver usando armadura:

* CA = 10 + Destreza + Constituição

Você pode usar escudo normalmente.

---

### **ATAQUE IMPRUDENTE (Nível 2)**

No primeiro ataque do turno, você pode decidir atacar de forma imprudente.

* Ganha vantagem nas jogadas de ataque Corpo a Corpo usando Força
* Ataques contra você têm vantagem até seu próximo turno

---

### **SENTIDO DE PERIGO (Nível 2)**

Você tem vantagem em testes de resistência de Destreza.

Você não recebe esse benefício se estiver:

* incapacitado

---

### **ATAQUE EXTRA (Nível 5)**

Você pode atacar duas vezes, em vez de uma, sempre que usar a ação de ataque no seu turno.

---

### **MOVIMENTO RÁPIDO (Nível 5)**

Seu deslocamento aumenta em **+3 metros** enquanto não estiver usando armadura pesada.

---

### **BOTE INSTINTIVO (Nível 7)**

Como parte da Ação Bônus que você realiza para entrar em Fúria, você pode se mover até metade do seu Deslocamento.

---

### **INSTINTOS PRIMITIVOS (Nível 7)**

Você tem vantagem nas jogadas de iniciativa.

---

### **GOLPE BRUTAL (Nível 9)**

Se você usar Ataque Imprudente, pode renunciar à Vantagem em uma jogada de ataque com Força à sua escolha no seu turno. A jogada de ataque escolhida não deve ter Desvantagem. Se a jogada de ataque atingir o alvo, este sofre 1d10 pontos de dano adicional do mesmo tipo causado pela arma ou Ataque Desarmado, e você pode causar um efeito de Golpe Brutal à sua 
escolha. Você tem as seguintes opções de efeito:

* **Golpe Debilitador:** O Deslocamento do alvo é reduzido em 4,5 metros até o início do seu próximo turno. Um alvo pode ser afetado por apenas um Golpe Debilitador de cada vez — o mais recente.

* **Golpe Poderoso:** O alvo é empurrado 4,5 metros diretamente para longe de você. Em seguida, você pode se mover até metade do seu Deslocamento diretamente em direção ao alvo sem provocar Ataques de Oportunidade.

---

### **FÚRIA IMPLACÁVEL (Nível 11)**

Se cair a 0 PV enquanto estiver em fúria:

* pode fazer um teste de resistência de Constituição (CD 10)

Se passar:

* fica com **(2 * Nível) de PV** em vez de cair

A CD aumenta em +5 a cada uso após o primeiro, até descanso longo.

---

### **GOLPE BRUTAL FORTALECIDO (Nível 13)**

Você aperfeiçoou novas formas de atacar com ferocidade. Os seguintes efeitos agora estão entre suas opções de Golpe Brutal:

* **Golpe Atordoante:** O alvo tem Desvantagem na próxima salvaguarda que realizar e não pode realizar Ataques de Oportunidade até o início do seu próximo turno.

* **Golpe Destruidor:** Antes do início do seu próximo turno, a próxima jogada de ataque realizada por outra criatura contra o alvo recebe um bônus de +5. Uma jogada de ataque só pode receber um bônus de Golpe Destruidor por vez.

---

* cair inconsciente

### **ARAUTO DA FAUNA (Nível 3)**

Você pode conjurar as magias Falar com Animais e Sentido Feral, mas apenas como Rituais. Sabedoria é seu atributo de conjuração para elas.

---

### **FÚRIA DOS SELVAGENS (Nível 3)**

Sua Fúria se conecta ao poder primitivo dos animais. Sempre que você ativar sua Fúria, adquire uma das seguintes opções à sua escolha.

* **Águia:** Ao ativar sua Fúria, você pode executar as ações Correr e Desengajar como parte dessa Ação Bônus. Enquanto sua Fúria estiver ativa, você pode executar uma Ação Bônus para realizar ambas as ações.

* **Lobo:** Enquanto sua Fúria estiver ativa, seus aliados têm Vantagem em jogadas de ataque contra qualquer inimigo seu a até 1,5 metro de você.

* **Urso:** Enquanto em Fúria, você tem Resistência a todos os tipos de dano, exceto Energético, Necrótico, Psíquico e Radiante.

---

### **ASPECTO DOS SELVAGENS (Nível 6)**

Você recebe uma das seguintes opções à sua escolha. Sempre que completar um Descanso Longo, você pode alterar sua escolha.

* **Coruja:** Você tem Visão no Escuro com um alcance de 18 metros. Se você já tem Visão no Escuro, seu alcance aumenta em 18 metros.

* **Pantera:** Você tem um Deslocamento de Escalada igual ao seu Deslocamento.

* **Salmão:** Você tem um Deslocamento de Natação igual ao seu Deslocamento.

---

### **ARAUTO DA NATUREZA (Nível 10)**

Você pode conjurar a magia Comunhão com a Natureza, mas apenas como um Ritual. Sabedoria é seu atributo de conjuração para isso.

---

### **PRESENÇA INTIMIDANTE (Nível 14)**

Sempre que você ativar sua Fúria, recebe uma das seguintes opções à sua escolha.

* **Carneiro:** Enquanto sua Fúria estiver ativa, você pode impor a condição Caído em uma criatura Grande ou menor quando você a atinge com um ataque corpo a corpo.

* **Falcão:** Enquanto sua Fúria estiver ativa, você tem um Deslocamento de Voo igual ao seu Deslocamento se não estiver usando nenhuma armadura.

* **Leão:** Enquanto sua Fúria estiver ativa, qualquer um de seus inimigos a até 1,5 metro de você tem Desvantagem em jogadas de ataque contra alvos que não sejam você ou outro Bárbaro que tenha essa opção ativa.

### **MÃOS CURATIVAS (Raça)**

Você executa uma ação Usar Magia, toca uma criatura e joga um número de d4s igual ao seu Bônus de Proficiência. A criatura restaura número de Pontos de Vida igual ao total jogado. Após usar esse traço, você não pode usá-lo novamente até completar um Descanso Longo.

### **REVELAÇÃO CELESTIAL (Raça)**

No nível 3 de personagem, você pode se transformar como uma Ação Bônus usando uma das opções abaixo (escolha a opção cada vez que você se transformar). A transformação se mantém por 1 minuto ou até você a encerrar (nenhuma ação é necessária). Uma vez que você se transforma, não pode fazê-lo novamente até completar um Descanso Longo. Uma vez em cada um dos seus turnos, até que a transformação termine, você pode infligir dano adicional a um alvo ao causar dano a ele com um ataque ou uma magia. O dano adicional é igual ao seu Bônus de Proficiência, e o tipo de dano adicional é Necrótico para Manto Necrótico ou Radiante para Asas Celestiais e Transfiguração Radiante. Aqui estão as opções de transformação:

* **Asas Celestiais:** Duas asas espectrais brotam em suas costas temporariamente. Até que a transformação se encerre, você tem um Deslocamento de Voo igual ao seu Deslocamento.

* **Manto Necrótico:** Seus olhos se tornam brevemente poças de escuridão, e asas que não voam brotam em suas costas temporariamente. Criaturas que não sejam seus aliados a até 3 metros de você devem ser bem-sucedidas em uma salvaguarda de Carisma (CD 8 + seu modificador de Carisma e seu Bônus de Proficiência) ou têm a condição Amedrontado até o final do seu próximo turno.

* **Transfiguração Radiante:** Luz abrasadora irradia temporariamente de seus olhos e boca. Pela duração da transformação, você emite Luz Plena em um raio de 3 metros e Meia-luz por mais 3 metros, e no fim de cada um de seus turnos, cada criatura a até 3 metros de você sofre dano Radiante igual ao seu Bônus de Proficiência

### **EXTRERMINADOR DE CONJURADORES** (Nível 4)

Aumenta o valor de Destreza em 1 e ganha os seguintes benefícios:

* **Quebrador de Concentração:** Quando você causa dano a uma criatura que está se concentrando em uma magia, ela tem Desvantagem na salvaguarda que realiza para manter a Concentração
* **Resguardo Mental:** Se falhar em uma salvaguarda de Inteligência, Sabedoria ou Carisma, em vez disso você escolhe ser bem-sucedido. Após usar esse benefício, você não pode usá-lo novamente até completar um Descanso Curto ou Longo.

### **MESTRE EM ARMAS DE HASTE** (Nível 8)

Aumenta o valor de Destreza em 1 e ganha os seguintes benefícios:

* **Golpe de Haste:** Você pode usar uma Ação Bônus para realizar um ataque Corpo a Corpo e o dado de dano da arma é d4.
* **Golpe Reativo:** Com uma reação, você pode atacar uma criatura que entra no seu alcance.

---

### **MESTRE EM ARMAS GRANDES** (Nível 12)

Aumenta o valor de Destreza em 1 e ganha os seguintes benefícios:

* **Maestria em Armas Pesadas:** Você consegue causar dano adicional com uma arma pesada igual ao seu bônus de proficiência.
* **Cortar:**  Imediatamente após obter um Acerto Crítico ou reduzir uma criatura a 0 Pontos de Vida com uma arma Corpo a Corpo, você pode realizar um ataque com a mesma arma como uma Ação Bônus.

---

### **AGRESSOR** (Nível 17)

Aumenta o valor de Destreza em 1 e ganha os seguintes benefícios:

* **Corrida Aprimorada:** Quando você executa a ação Correr, seu Deslocamento aumenta em 3 metros para esta ação.
* **Ataque em Investida:** Se você se mover pelo menos 3 metros em linha reta em direção a um alvo imediatamente antes de atingi-lo com uma jogada de ataque Corpo a Corpo como parte da ação Atacar, escolha um dos seguintes efeitos: obter um bônus de +1d8 na joga da de dano do ataque ou empurrar o alvo até 3 metros se ele não for um tamanho maior que você. Você pode usar esse benefício apenas uma vez em cada um dos seus turnos.

---

### **DÁDIVA DA EVOLUÇÃO**

Até o maior dos monstros já começou comouma criatura frágil e incapaz, então supere-se e torne-se algo melhor.

* **Evolução gradual:** A partir do segundo turno, você pode gastar 1 ponto de sanidade em uma ação livre para aumentar sua proficiência bônus em 1 ponto. Você só pode fazer isso uma vez por rodada e este bônus dura até o fim da cena.
* **Sobrecarga divina:** No seu primeiro turno em cenas de combates, você sofre acertos crı́ticos para qualquer coisa que lhe acertar. Se tiver “Predação”, você pode usar sua reação para impedir um crı́tico que te acerte na primeira rodada.

---

### **DÁDIVA DA PREDAÇÃO**

Um dia a presa se torna o caçador... sobreviva até este dia chegar.

* **Predação:** Torna-se capaz de conjurar “marca docaçador” usando 1 ponto de sanidade e uma açãobônus. Posteriormente, também pode usar 1 ponto desanidade e uma ação bônus para aumentar o dano da marca em um passo. Caso a criatura marcada morra,você pode escolher um novo alvo da marca, que sofrerá com o mesmo passo de dano que o anterior + 1 passo. Isso não gasta ação ou pontos de sanidade. A magia não é de concentração para você.
* **Sobrecarga divina:** Você sempre está sobre efeito de marca do caçador para as criaturas mais poderosas. Em outro caso, você tenha “evolução gradual”, você perde a marca no final do terceiro turno.

**Runa de MUI**

Você tem uma item chamado "Runa de MUI" que pode ser usada para invocar a Mui como uma ação bônus. Ele é um artefato de classe X que pesa 1 kilo. Tem somente 1 uso por descanço longo.

---

```python
# 🎲 **ROLAGEM DE DANO MIM**

def forca_numero(msg):
    while True:
        valor = input(msg)
        if valor.isnumeric():
            return int(valor)
        print("Formato inválido!")

def forca_opcao(msg, opcoes):
    while True:
        valor = input(msg).lower()
        if valor in opcoes:
            return valor
        print("Opção inválida!")

opcoes = ['s', 'n']

nivel = forca_numero("Qual é o seu nível? ")
mod_forca = forca_numero("Modificador de Força: ")
critico = forca_opcao("Foi crítico? (s/n): ", opcoes)

if nivel < 9:
    furia = 2
elif nivel < 16:
    furia = 3
else:
    furia = 4

if nivel < 5:
    prof = 2
elif nivel < 9:
    prof = 3
elif nivel < 13:
    prof = 4
elif nivel < 17:
    prof = 5
else:
    prof = 6

arma = "1d10"

dados = []
bonus = mod_forca + furia

dados.append(arma)

if nivel >= 3:
    evolucao = forca_opcao("Usou a dádiva evolução? (s/n): ", opcoes)
    if evolucao == 's':
        qtd_evolucao = forca_numero("Quantas vezes?")
        if qtd_evolucao == 1:
            prof += 1
        elif qtd_evolucao == 2:
            prof += 2
        elif qtd_evolucao == 3:
            prof += 3
        elif qtd_evolucao == 4:
            prof += 4
        elif qtd_evolucao >= 5:
            prof += 5

    predacao = forca_opcao("Usou a dádiva predação? (s/n): ", opcoes)

    if predacao == 's':
        qtd_predacao = forca_numero("Quantas vezes?")
        if qtd_predacao == 1:
            dados.append("1d6")
        elif qtd_predacao == 2:
            dados.append("1d8")
        elif qtd_predacao == 3:
            dados.append("1d10")
        elif qtd_predacao == 4:
            dados.append("1d12")
        elif qtd_predacao >= 5:
            dados.append("2d12")

    imprudente = forca_opcao("Ataque Imprudente? (s/n): ", opcoes)

    if imprudente == 's':
        usar_brutal = forca_opcao("Usar Golpe Brutal? (s/n): ", opcoes)
        if usar_brutal == 's':
            if nivel >= 17:
                dados.append("2d10")
            elif nivel >= 9:
                dados.append("1d10")

        transformacao = forca_opcao("Usar Revelação Celestial? (s/n): ", opcoes)
        if transformacao == 's':
            bonus += prof
        
        perto = forca_opcao("Usar Transfiguração Radiante? (s/n): ", opcoes)
        if perto == 's':
            bonus += prof

if nivel >= 8:
    haste = forca_opcao("Atacar com a haste? (s/n): ", opcoes)
    if haste == 's':
        dados[0] = "1d4"

if nivel >= 12:
    usar_investida = forca_opcao("Usar Investida? (s/n): ", opcoes)
    if usar_investida == 's':
        dados.append("1d8")
    bonus += prof 

if critico == 's':
    dados += dados  

expressao = " + ".join(dados) + f" + {bonus}"

print(f"🎲 Copiar para o bot: {expressao}")
```
