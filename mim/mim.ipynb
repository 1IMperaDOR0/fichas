## 📦 **INVENTÁRIO**

### ⚔️ **ARMAS**

**Lança de Montaria**  
* Alcance: 3 m
* Dano: 1d10 perfurante  
* Propriedades: Versátil (se montado), Pesada, Extensão e Derrubar
  * **Derrubar:** Se você atingir uma criatura com esta arma, você pode forçar a criatura a realizar uma salvaguarda de Constituição (CD 8 mais o modificador de atributo usado para realizar a jogada de ataque e seu Bônus de Proficiência). Se falhar, a criatura tem a condição Caído.

```python
# 🎲 **ROLAGEM DE DANO**

def forca_numero(msg):
    valor = input(msg)
    if not valor.isnumeric():
        print("Formato inválido!")
        valor = forca_numero(msg)
    return int(valor)

def forca_opcao(msg, opcoes):
    opcao = input(msg)
    if not opcao in opcoes:
        print("Opção inválida!")
        opcao = forca_opcao(msg, opcoes)
    return opcao

opcoes = ['s', 'n']

nivel = forca_numero("Qual é o seu nível?")

if nivel < 9:
    furia = 2
elif nivel < 16:
    furia = 3
else:
    furia = 4

modificador_forca = forca_numero("Qual o seu modificador de Força?") 

critico = forca_opcao("O acerto foi Crítico?", opcoes)

lanca_montaria = "1d10"

frenesi = 0

golpe_brutal = 0

bonus_proficiente = 0

investida = 0

if nivel >= 3:
    ataque_imprudente = forca_opcao("Usou a habilidade Ataque Imprudente?", opcoes)
    if ataque_imprudente == 's':
        precisa = forca_opcao("Este ataque pode usar a habilidade Frenesi?", opcoes)
        if precisa == 's':
            frenesi = f"{furia}d6"
        precisa = forca_opcao("Este ataque pode usar a habilidade Golpe Brutal", opcoes)
        if precisa == 's':
            if nivel >= 9 and nivel < 17:
                golpe_brutal = "1d10"
            else:
                golpe_brutal = "2d10"

if nivel > 7:
    haste = forca_opcao("Está atacando com a haste?", opcoes)
    if haste == 's':
        lanca_montaria = "1d4"

if nivel > 11:
    if nivel < 17:
        bonus_proficiente = 5
    else:
        bonus_proficiente = 6

    precisa = forca_opcao("Este ataque pode usar a habilidade Investida?", opcoes)
    if precisa == 's':
        investida = "1d8"

dados = []
bonus = modificador_forca + furia + bonus_proficiente

dados.append(lanca_montaria)

if frenesi != 0:
    dados.append(frenesi)

if golpe_brutal != 0:
    dados.append(golpe_brutal)

if investida != 0:
    dados.append(investida)

if critico == 's':
    dados += dados 

dano = " + ".join(dados) + f" + {bonus}"

print(dano)
```
